# 01 OAR 检索预计算

本 notebook 对原始短语执行单路 OAR（Ontology-Aligned Retrieval）并保存 Top-20。HPO 的 text-embedding-3-large 查询向量经过训练后的 3072→3072 无偏置线性映射和 L2 归一化后，直接在专用 OAR cosine Chroma 中检索。

主流程仅包含单路 OAR cosine 检索。其他 embedding 模型和 MONDO 使用各自原始 Dense cosine 检索，仅用于切换模型或 ontology 时的对照运行。

输出为 `results/precompute/*.xlsx`。每条样本占一行，完整 Top-20 保存在 JSON 列中，并将 20 个候选另行展开。

六个固定 HPO 原始 embedding collection 的构建、断点续建和一致性校验也集中在本 notebook，通过 `HPO_CHROMA_BUILD_MODELS` 显式启用。

## 3. 数据结构与基础工具


```python
# 中文注释：保存一个 ontology 概念的名称、同义词和定义。
@dataclass(frozen=True)
class OntologyTerm:
    ontology_id: str
    name: str
    synonyms: tuple[tuple[str, str], ...]
    definition: str


# 中文注释：保存词法和 Dense 索引中的名称或同义词表面形式及其概念信息。
@dataclass(frozen=True)
class SurfaceForm:
    text: str
    ontology_id: str
    preferred_name: str
    definition: str
    synonym_type: str


# 中文注释：保存按 cosine 排序并折叠到 concept 后的单路 Dense 候选。
@dataclass(frozen=True)
class RankedCandidate:
    ontology_id: str
    dense_score: float
    matched_surface: str


# 中文注释：读取 JSON 配置但不打印 API key。
def load_config(path: Path) -> dict[str, Any]:
    with path.open("r", encoding="utf-8-sig") as handle:
        return json.load(handle)


# 中文注释：计算文件哈希，用于缓存失效和运行记录。
def file_sha256(path: Path) -> str:
    digest = hashlib.sha256()
    with path.open("rb") as handle:
        for block in iter(lambda: handle.read(1024 * 1024), b""):
            digest.update(block)
    return digest.hexdigest()


# 中文注释：把模型名转换成稳定、可用于文件名和 Chroma collection 的短标签。
def safe_tag(value: str, limit: int = 48) -> str:
    tag = re.sub(r"[^A-Za-z0-9._-]+", "_", value).strip("._-").lower()
    return (tag or "model")[:limit]


# 中文注释：统一 Unicode、大小写和空白，词法召回不会改写原始查询值。
def normalize_text(value: str) -> str:
    return re.sub(r"\s+", " ", str(value or "").strip().lower())


# 中文注释：解析 MONDO OBO JSONL 中带引号和 scope 的同义词字符串。
def parse_mondo_synonym(value: str) -> str:
    match = re.match(r'^"((?:\\.|[^"\\])*)"', str(value or ""))
    if not match:
        return str(value or "").strip()
    return json.loads(f'"{match.group(1)}"')


# 中文注释：读取固定版本 HPO 或 MONDO 术语并统一成相同的数据结构。
def load_ontology(ontology_key: str) -> tuple[dict[str, OntologyTerm], str]:
    spec = ONTOLOGIES[ontology_key]
    terms: dict[str, OntologyTerm] = {}
    with spec["path"].open("r", encoding="utf-8-sig") as handle:
        for line in handle:
            if not line.strip():
                continue
            item = json.loads(line)
            if ontology_key == "hpo":
                if item.get("is_obsolete"):
                    continue
                ontology_id = str(item["HPOid"])
                synonyms = tuple(
                    (
                        str(entry.get("text") or ""),
                        str(entry.get("predicate") or "hasRelatedSynonym"),
                    )
                    for entry in item.get("synonyms", [])
                )
                definition = str(item.get("definition") or "")
            else:
                ontology_id = str(item["MONDO_id"])
                synonyms = tuple(
                    (parse_mondo_synonym(entry), "hasExactSynonym")
                    for entry in item.get("synonym", [])
                )
                definition = str(item.get("description") or "").strip('"')
            terms[ontology_id] = OntologyTerm(
                ontology_id=ontology_id,
                name=str(item["name"]),
                synonyms=tuple((text, predicate) for text, predicate in synonyms if text),
                definition=definition,
            )
    return terms, file_sha256(spec["path"])


# 中文注释：HPO 每个概念内按原顺序保留首次出现的 surface 文本；preferred name 优先。
def build_surfaces(terms: dict[str, OntologyTerm], ontology_key: str) -> list[SurfaceForm]:
    surfaces: list[SurfaceForm] = []
    for term in terms.values():
        entries = [(term.name, "preferred_label"), *term.synonyms]
        seen_texts: set[str] = set()
        for text, synonym_type in entries:
            text_value = str(text).strip()
            if not text_value or (ontology_key == "hpo" and text_value in seen_texts):
                continue
            if ontology_key == "hpo":
                seen_texts.add(text_value)
            surfaces.append(SurfaceForm(
                text=text_value,
                ontology_id=term.ontology_id,
                preferred_name=term.name,
                definition=term.definition,
                synonym_type=synonym_type,
            ))
    return surfaces


# 中文注释：解析当前主 Gold ID及其官方历史可接受 ID；MONDO 默认只接受主 Gold ID。
def accepted_gold_ids(sample: dict[str, Any], spec: dict[str, Any]) -> tuple[str, ...]:
    gold_id = str(sample[spec["gold_col"]]).strip()
    if spec["ontology"] != "hpo":
        return (gold_id,)
    raw_value = sample.get("Accepted_HPO_ids", "")
    try:
        payload = json.loads(str(raw_value))
    except json.JSONDecodeError as exc:
        raise ValueError(f"Accepted_HPO_ids 不是合法 JSON: {raw_value}") from exc
    if not isinstance(payload, list):
        raise ValueError("Accepted_HPO_ids 必须是 JSON 数组")
    accepted = tuple(dict.fromkeys(str(value).strip() for value in payload))
    if gold_id not in accepted:
        raise ValueError(f"当前 Gold ID 不在 Accepted_HPO_ids 中: {gold_id}")
    pattern = re.compile(ONTOLOGIES[spec["ontology"]]["id_pattern"])
    invalid = [value for value in accepted if not pattern.fullmatch(value)]
    if invalid:
        raise ValueError(f"Accepted_HPO_ids 包含非法 ID: {invalid[0]}")
    return accepted


# 中文注释：读取测试集并生成稳定 sample_id，同时验证 Gold 和 accepted ID。
def load_samples(dataset_key: str, max_samples: int | None = None) -> tuple[pd.DataFrame, dict[str, Any]]:
    spec = DATASETS[dataset_key]
    frame = pd.read_excel(spec["path"], sheet_name=spec["sheet"], dtype=str).fillna("")
    required = {spec["query_col"], spec["preferred_col"], spec["gold_col"], *spec["metadata_cols"]}
    if spec["ontology"] == "hpo":
        required.add("Accepted_HPO_ids")
    missing = required - set(frame.columns)
    if missing:
        raise KeyError(f"数据集缺少列: {sorted(missing)}")
    if max_samples is not None:
        frame = frame.head(int(max_samples)).copy()
    frame.insert(0, "sample_id", np.arange(1, len(frame) + 1, dtype=int))
    frame.insert(1, "source_excel_row", np.arange(2, len(frame) + 2, dtype=int))
    pattern = re.compile(ONTOLOGIES[spec["ontology"]]["id_pattern"])
    invalid = frame.loc[~frame[spec["gold_col"]].map(lambda value: bool(pattern.fullmatch(str(value).strip())))]
    if not invalid.empty:
        raise ValueError(f"发现非单一或非法 Gold ID，首个 source_excel_row={invalid.iloc[0]['source_excel_row']}")
    for sample in frame.to_dict(orient="records"):
        accepted_gold_ids(sample, spec)
    return frame, spec
```

## 4. Embedding 后端、OAR 投影与 Dense 索引

```python
# 中文注释：统一封装 OpenAI-compatible API 和本地 sentence-transformers 模型。
class EmbeddingBackend:
    # 中文注释：解析模型、并发、维度和 API 地址，不输出凭证。
    def __init__(
        self,
        backend: str,
        model: str,
        config: dict[str, Any],
        dimensions: int | None,
        max_concurrency: int,
        batch_size: int,
    ):
        self.backend = backend
        self.model = model
        self.dimensions = dimensions
        self.max_concurrency = int(max_concurrency)
        self.batch_size = int(batch_size)
        self._thread_local = threading.local()
        if backend == "sentence_transformers":
            self.local_model = SentenceTransformer(model)
        elif backend == "openai":
            self.local_model = None
            self.api_key = config["api_key"]
            self.base_url = config.get("base_url") or None
        else:
            raise ValueError("内部 embedding 调用方式只能是 openai 或 sentence_transformers")

    # 中文注释：为当前线程创建独立 OpenAI client，避免并发共享连接状态。
    def _client(self) -> OpenAI:
        if not hasattr(self._thread_local, "client"):
            kwargs = {"api_key": self.api_key}
            if self.base_url:
                kwargs["base_url"] = self.base_url
            self._thread_local.client = OpenAI(**kwargs)
        return self._thread_local.client

    # 中文注释：对一个远程批次调用 embedding API 并返回 float32 数组。
    def _encode_remote_batch(self, texts: Sequence[str]) -> np.ndarray:
        kwargs: dict[str, Any] = {"model": self.model, "input": list(texts)}
        if self.dimensions is not None:
            kwargs["dimensions"] = int(self.dimensions)
        response = self._client().embeddings.create(**kwargs)
        ordered = sorted(response.data, key=lambda item: item.index)
        return np.asarray([item.embedding for item in ordered], dtype=np.float32)

    # 中文注释：按配置后端批量编码文本，并对向量做 L2 归一化。
    def encode(self, texts: Sequence[str]) -> np.ndarray:
        values = [str(text) for text in texts]
        if not values:
            return np.empty((0, 0), dtype=np.float32)
        if self.backend == "sentence_transformers":
            vectors = self.local_model.encode(
                values,
                batch_size=self.batch_size,
                show_progress_bar=False,
                convert_to_numpy=True,
                normalize_embeddings=True,
            ).astype(np.float32)
            return vectors
        batches = [values[index:index + self.batch_size] for index in range(0, len(values), self.batch_size)]
        with ThreadPoolExecutor(max_workers=self.max_concurrency) as executor:
            parts = list(executor.map(self._encode_remote_batch, batches))
        vectors = np.vstack(parts)
        norms = np.linalg.norm(vectors, axis=1, keepdims=True)
        return vectors / np.maximum(norms, 1e-12)


# 中文注释：根据 ontology 固定版本解析 Chroma 持久化目录，允许参数显式覆盖。
def resolve_chroma_db_path(ontology_key: str) -> Path:
    if CHROMA_DB_DIR is not None:
        candidate = Path(CHROMA_DB_DIR).expanduser()
        return candidate.resolve() if candidate.is_absolute() else (ROOT / candidate).resolve()
    defaults = {
        "hpo": ROOT / "chroma_db_hpo_260623",
        "mondo": ROOT / "chroma_db_mondo_260706",
    }
    return defaults[ontology_key].resolve()


# 中文注释：生成新 surface collection 的稳定名称，并优先映射仓库已有的 HPO collection。
def dense_collection_names(ontology_key: str, embedding_model: str) -> tuple[str | None, str]:
    preferred = {
        ("hpo", "text-embedding-3-large"): "hpo_text_embedding_3_large_20260623",
        ("hpo", "text-embedding-3-small"): "hpo_text_embedding_3_small_20260623",
        ("hpo", "FremyCompany/BioLORD-2023"): "hpo_biolord_20260623",
        ("hpo", "dmis-lab/biobert-base-cased-v1.2"): "hpo_biobert_base_cased_v1_2_mean_pooling_20260623",
    }.get((ontology_key, embedding_model))
    release_tag = ONTOLOGIES[ontology_key]["release"].replace("-", "")
    model_hash = hashlib.sha256(embedding_model.encode("utf-8")).hexdigest()[:8]
    generated = safe_tag(f"{ontology_key}_{safe_tag(embedding_model, 30)}_{release_tag}_{model_hash}", limit=63)
    return preferred, generated


# 中文注释：生成包含 surface ID、文本、类型和顺序的稳定 catalog fingerprint。
def surface_catalog_fingerprint(surfaces: Sequence[SurfaceForm]) -> str:
    synonym_ordinals: dict[str, int] = defaultdict(int)
    payload: list[dict[str, Any]] = []
    for index, surface in enumerate(surfaces):
        ordinal = None
        if surface.synonym_type != "preferred_label":
            ordinal = synonym_ordinals[surface.ontology_id]
            synonym_ordinals[surface.ontology_id] += 1
        payload.append({
            "index": index,
            "text": surface.text,
            "hpo_id": surface.ontology_id,
            "preferred_label": surface.preferred_name,
            "kind": surface.synonym_type,
            "synonym_ordinal": ordinal,
        })
    encoded = json.dumps(payload, ensure_ascii=True, sort_keys=True, separators=(",", ":"))
    return hashlib.sha256(encoded.encode("utf-8")).hexdigest()


# 中文注释：严格校验 ontology、manifest、数量、模型、维度和 cosine 配置。
def collection_matches(
    collection: Any,
    ontology_key: str,
    ontology_hash: str,
    embedding_model: str,
    preferred_name: str | None,
    requested_dimensions: int | None,
    expected_surface_count: int,
    catalog_fingerprint: str,
) -> bool:
    metadata = collection.metadata or {}
    stored_model = str(metadata.get("embedding_model") or metadata.get("model_name") or "")
    if stored_model != embedding_model:
        return False
    stored_dimensions = metadata.get("embedding_dimensions") or metadata.get("dimension")
    if requested_dimensions is not None and stored_dimensions not in (None, ""):
        if int(stored_dimensions) != int(requested_dimensions):
            return False
    stored_hash = str(metadata.get("hpo_jsonl_sha256") or metadata.get("ontology_sha256") or "")
    if stored_hash != ontology_hash:
        return False
    stored_fingerprint = str(metadata.get("catalog_fingerprint") or metadata.get("surface_manifest_sha256") or "")
    if stored_fingerprint != catalog_fingerprint:
        return False
    if int(metadata.get("expected_surface_count") or 0) != int(expected_surface_count):
        return False
    if collection.count() != int(expected_surface_count):
        return False
    configuration = collection.configuration_json or {}
    configured_space = str((configuration.get("hnsw") or {}).get("space") or metadata.get("hnsw_space") or "").lower()
    if configured_space != "cosine":
        return False
    ontology_matches = (
        str(metadata.get("ontology") or "").lower() == ontology_key
        or collection.name.lower().startswith(f"{ontology_key}_")
    )
    return ontology_matches


# 中文注释：列出客户端 collection 并统一处理 Chroma 新旧版本返回名称或对象的差异。
def list_dense_collections(client: Any) -> list[Any]:
    collections: list[Any] = []
    for item in client.list_collections():
        name = item if isinstance(item, str) else item.name
        collections.append(client.get_collection(str(name)))
    return collections


# 中文注释：OAR 索引由最终调优脚本一次性建立；主流程严格校验模型、catalog、数量和 cosine 配置后复用。
def prepare_oar_collection(
    surfaces: Sequence[SurfaceForm],
    oar_runtime: Any,
) -> tuple[Any, dict[str, Any]]:
    db_path = (ROOT / DEFAULT_OAR_CHROMA_PATH).resolve()
    if not db_path.exists():
        raise FileNotFoundError(f"找不到最终 OAR Chroma，请先完成 OAR 投影训练和索引构建: {db_path}")
    client = chromadb.PersistentClient(path=str(db_path))
    expected_catalog = surface_catalog_fingerprint(surfaces)
    matches = []
    for collection in list_dense_collections(client):
        metadata = collection.metadata or {}
        configuration = collection.configuration_json or {}
        space = str((configuration.get("hnsw") or {}).get("space") or metadata.get("hnsw_space") or "").lower()
        if (
            metadata.get("projection_family") == oar_runtime.collection_metadata().get("projection_family")
            and metadata.get("projection_model_sha256") == oar_runtime.model_sha256
            and metadata.get("catalog_fingerprint") == expected_catalog
            and int(metadata.get("projection_output_dimension") or 0) == 3072
            and collection.count() == len(surfaces)
            and space == "cosine"
        ):
            matches.append(collection)
    if len(matches) != 1:
        raise RuntimeError(f"最终 OAR Chroma 匹配数量应为 1，实际为 {len(matches)}: {db_path}")
    collection = matches[0]
    info = {
        "chroma_db_dir": str(db_path),
        "dense_collection_name": collection.name,
        "dense_collection_count": collection.count(),
        "dense_index_unit": "surface",
        "dense_collection_action": "reused_final_oar",
        "dense_catalog_version": (collection.metadata or {}).get("catalog_version", ""),
        "oar_run_id": oar_runtime.run_id,
        "oar_model_sha256": oar_runtime.model_sha256,
    }
    print(f"OAR collection reused: {collection.name}; count={collection.count()}; path={db_path}")
    return collection, info


# 中文注释：在固定目录中复用匹配 collection；不存在时按表面形式创建并支持中断续建。
def prepare_dense_collection(
    ontology_key: str,
    ontology_hash: str,
    surfaces: Sequence[SurfaceForm],
    embedder: EmbeddingBackend,
) -> tuple[Any, dict[str, Any]]:
    db_path = resolve_chroma_db_path(ontology_key)
    db_path.mkdir(parents=True, exist_ok=True)
    client = chromadb.PersistentClient(path=str(db_path))
    preferred_name, generated_name = dense_collection_names(ontology_key, embedder.model)
    catalog_fingerprint = surface_catalog_fingerprint(surfaces)
    collections = list_dense_collections(client)
    matches = [collection for collection in collections if collection_matches(
        collection,
        ontology_key,
        ontology_hash,
        embedder.model,
        preferred_name,
        embedder.dimensions,
        len(surfaces),
        catalog_fingerprint,
    )]
    matches.sort(key=lambda collection: (collection.name != preferred_name, collection.name))

    if REBUILD_DENSE_INDEX:
        target_name = matches[0].name if matches else (preferred_name or generated_name)
        for collection in matches:
            client.delete_collection(collection.name)
        matches = []
    else:
        target_name = preferred_name or generated_name
        for collection in matches:
            build_status = str((collection.metadata or {}).get("build_status") or "")
            if collection.count() > 0 and build_status != "building":
                info = {
                    "chroma_db_dir": str(db_path),
                    "dense_collection_name": collection.name,
                    "dense_collection_count": collection.count(),
                    "dense_index_unit": "surface",
                    "dense_collection_action": "reused",
                    "dense_catalog_version": (collection.metadata or {}).get("catalog_version", ""),
                }
                print(f"Dense collection reused: {collection.name}; count={collection.count()}; path={db_path}")
                return collection, info
        if matches:
            target_name = matches[0].name

    catalog_version = (
        f"hp/releases/{ONTOLOGIES[ontology_key]['release']}"
        if ontology_key == "hpo"
        else f"mondo/releases/{ONTOLOGIES[ontology_key]['release']}"
    )
    metadata = {
        "ontology": ontology_key,
        "ontology_sha256": ontology_hash,
        "ontology_release": ONTOLOGIES[ontology_key]["release"],
        "catalog_version": catalog_version,
        "embedding_backend": embedder.backend,
        "embedding_model": embedder.model,
        "expected_surface_count": len(surfaces),
        "surface_manifest_sha256": catalog_fingerprint,
        "catalog_fingerprint": catalog_fingerprint,
        "distance_metric": "cosine",
        "hnsw_space": "cosine",
        "index_unit": "surface",
        "build_status": "building",
    }
    collection = next((item for item in matches if item.name == target_name), None)
    if collection is None:
        try:
            collection = client.get_collection(target_name)
        except Exception:
            collection = client.create_collection(target_name, metadata=metadata, configuration={"hnsw": {"space": "cosine"}})
    current_metadata = collection.metadata or {}
    if collection.count() and not collection_matches(
        collection, ontology_key, ontology_hash, embedder.model, preferred_name, embedder.dimensions, len(surfaces), catalog_fingerprint
    ):
        raise ValueError(f"Chroma collection 名称冲突且 metadata 不匹配: {target_name}")

    existing_ids = set(collection.get(include=[]).get("ids", [])) if collection.count() else set()
    initial_count = len(existing_ids)
    actual_dimensions: int | None = None
    for start in range(0, len(surfaces), INDEX_BATCH_SIZE):
        indexes = [
            index for index in range(start, min(start + INDEX_BATCH_SIZE, len(surfaces)))
            if f"surface_{index:07d}" not in existing_ids
        ]
        if not indexes:
            continue
        batch = [surfaces[index] for index in indexes]
        documents = [surface.text for surface in batch]
        embeddings = embedder.encode(documents)
        actual_dimensions = int(embeddings.shape[1])
        id_key = "hpo_id" if ontology_key == "hpo" else "mondo_id"
        collection.add(
            ids=[f"surface_{index:07d}" for index in indexes],
            documents=documents,
            metadatas=[{
                "ontology_id": surface.ontology_id,
                id_key: surface.ontology_id,
                "canonical_name": surface.preferred_name,
                "preferred_name": surface.preferred_name,
                "definition": surface.definition,
                "synonym_type": surface.synonym_type,
                "catalog_version": catalog_version,
            } for surface in batch],
            embeddings=embeddings.tolist(),
        )
        if start == 0 or (start // INDEX_BATCH_SIZE + 1) % 25 == 0:
            print(f"Dense surface index: {collection.count()}/{len(surfaces)}")

    final_count = collection.count()
    if final_count != len(surfaces):
        raise RuntimeError(f"Dense surface 索引不完整: expected={len(surfaces)}, actual={final_count}")
    final_metadata = dict(metadata)
    final_metadata["build_status"] = "complete"
    if actual_dimensions is not None:
        final_metadata["embedding_dimensions"] = actual_dimensions
    elif current_metadata.get("embedding_dimensions") or current_metadata.get("dimension"):
        final_metadata["embedding_dimensions"] = int(
            current_metadata.get("embedding_dimensions") or current_metadata.get("dimension")
        )
    collection.modify(metadata=final_metadata)
    action = "resumed_and_built" if initial_count else "built"
    info = {
        "chroma_db_dir": str(db_path),
        "dense_collection_name": collection.name,
        "dense_collection_count": final_count,
        "dense_index_unit": "surface",
        "dense_collection_action": action,
        "dense_catalog_version": catalog_version,
    }
    print(f"Dense collection {action}: {collection.name}; count={final_count}; path={db_path}")
    return collection, info


# 中文注释：使用已计算的查询向量检索 Dense 候选，不重复调用 embedding API。
def dense_search(collection: Any, query_vector: np.ndarray, top_k: int) -> list[RankedCandidate]:
    collection_count = collection.count()
    if collection_count <= 0 or top_k <= 0:
        return []
    requested = min(collection_count, max(int(top_k) * 4, int(top_k)))
    while True:
        result = collection.query(
            query_embeddings=[query_vector.tolist()],
            n_results=requested,
            include=["documents", "metadatas", "distances"],
        )
        documents = (result.get("documents") or [[]])[0]
        metadatas = (result.get("metadatas") or [[]])[0]
        distances = (result.get("distances") or [[]])[0]
        output: list[RankedCandidate] = []
        seen_ids: set[str] = set()
        for document, metadata, distance in zip(documents, metadatas, distances):
            metadata = metadata or {}
            ontology_id = str(
                metadata.get("ontology_id")
                or metadata.get("hpo_id")
                or metadata.get("mondo_id")
                or ""
            )
            if not ontology_id or ontology_id in seen_ids:
                continue
            seen_ids.add(ontology_id)
            output.append(RankedCandidate(
                ontology_id=ontology_id,
                dense_score=1.0 - float(distance),
                matched_surface=str(document),
            ))
            if len(output) >= int(top_k):
                return output
        if requested >= collection_count:
            return output
        requested = min(collection_count, requested * 2)
```

```python
# 中文注释：六个 HPO Dense collection 的固定身份和构建逻辑集中在主 notebook；不再依赖独立构建脚本。
@dataclass(frozen=True)
class HPOEmbeddingSpec:
    key: str
    model_name: str
    collection_name: str
    backend: str
    revision: str
    pooling: str
    dimension: int
    batch_size: int
    config_section: str | None = None


HPO_EMBEDDING_SPECS = (
    HPOEmbeddingSpec(
        "text_embedding_3_large", "text-embedding-3-large",
        "hpo_text_embedding_3_large_20260623", "openai",
        "provider-managed model ID (immutable revision unavailable)",
        "provider embedding endpoint", 3072, 512, "embedding",
    ),
    HPOEmbeddingSpec(
        "text_embedding_3_small", "text-embedding-3-small",
        "hpo_text_embedding_3_small_20260623", "openai",
        "provider-managed model ID (immutable revision unavailable)",
        "provider embedding endpoint", 1536, 128, "embedding_small",
    ),
    HPOEmbeddingSpec(
        "biolord", "FremyCompany/BioLORD-2023", "hpo_biolord_20260623",
        "sentence_transformers", "167aab527b238a50ca65224e6319215d2ff4fc9f",
        "model-native SentenceTransformer pooling", 768, 128,
    ),
    HPOEmbeddingSpec(
        "biobert", "dmis-lab/biobert-base-cased-v1.2",
        "hpo_biobert_base_cased_v1_2_mean_pooling_20260623",
        "transformers_mean_pooling", "67c9c25b46986521ca33df05d8540da1210b3256",
        "attention-mask mean pooling of last_hidden_state", 768, 128,
    ),
    HPOEmbeddingSpec(
        "pubmedbert", "microsoft/BiomedNLP-BiomedBERT-base-uncased-abstract-fulltext",
        "hpo_pubmedbert_abstract_fulltext_mean_pooling_20260623",
        "transformers_mean_pooling", "e1354b7a3a09615f6aba48dfad4b7a613eef7062",
        "attention-mask mean pooling of last_hidden_state", 768, 128,
    ),
    HPOEmbeddingSpec(
        "clinicalbert", "emilyalsentzer/Bio_ClinicalBERT",
        "hpo_clinicalbert_mean_pooling_20260623",
        "transformers_mean_pooling", "d5892b39a4adaed74b92212a44081509db72f87b",
        "attention-mask mean pooling of last_hidden_state", 768, 128,
    ),
)
HPO_SPECS_BY_KEY = {spec.key: spec for spec in HPO_EMBEDDING_SPECS}
HPO_SPECS_BY_MODEL = {spec.model_name: spec for spec in HPO_EMBEDDING_SPECS}


class PinnedEmbeddingBackend:
    # 中文注释：查询和 surface 构建共用同一 revision、pooling 和 L2 归一化实现。
    def __init__(self, spec: HPOEmbeddingSpec, config: dict[str, Any], max_concurrency: int):
        self.spec = spec
        self.backend = spec.backend
        self.model = spec.model_name
        self.revision = spec.revision
        self.pooling = spec.pooling
        self.dimensions = spec.dimension
        self.batch_size = spec.batch_size
        self.max_concurrency = max(1, int(max_concurrency))
        self._thread_local = threading.local()
        self._stats_lock = threading.Lock()
        self.stats = {
            "api_call_batches": 0,
            "api_text_count": 0,
            "api_prompt_tokens": 0,
            "api_total_tokens": 0,
            "local_processed_text_count": 0,
        }
        self.device = "cuda" if torch.cuda.is_available() else "cpu"
        self.local_model = None
        self.tokenizer = None
        self.transformer = None
        if spec.backend == "openai":
            self.api_key = config["api_key"]
            self.base_url = config.get("base_url") or None
        elif spec.backend == "sentence_transformers":
            self.local_model = SentenceTransformer(
                spec.model_name,
                revision=spec.revision,
                device=self.device,
                local_files_only=True,
            )
        elif spec.backend == "transformers_mean_pooling":
            self.tokenizer = AutoTokenizer.from_pretrained(
                spec.model_name, revision=spec.revision, local_files_only=True
            )
            self.transformer = AutoModel.from_pretrained(
                spec.model_name, revision=spec.revision, local_files_only=True
            ).to(self.device)
            self.transformer.eval()
        else:
            raise ValueError(f"不支持的固定 embedding backend: {spec.backend}")

    def _client(self) -> OpenAI:
        if not hasattr(self._thread_local, "client"):
            kwargs = {"api_key": self.api_key, "max_retries": 0}
            if self.base_url:
                kwargs["base_url"] = self.base_url
            self._thread_local.client = OpenAI(**kwargs)
        return self._thread_local.client

    def _encode_remote_batch(self, texts: Sequence[str]) -> np.ndarray:
        response = self._client().embeddings.create(model=self.model, input=list(texts))
        ordered = sorted(response.data, key=lambda item: item.index)
        usage = getattr(response, "usage", None)
        with self._stats_lock:
            self.stats["api_call_batches"] += 1
            self.stats["api_text_count"] += len(texts)
            self.stats["api_prompt_tokens"] += int(getattr(usage, "prompt_tokens", 0) or 0)
            self.stats["api_total_tokens"] += int(getattr(usage, "total_tokens", 0) or 0)
        return np.asarray([item.embedding for item in ordered], dtype=np.float32)

    def encode(self, texts: Sequence[str]) -> np.ndarray:
        values = [str(text) for text in texts]
        if not values:
            return np.empty((0, self.dimensions), dtype=np.float32)
        if self.backend == "openai":
            batches = [
                values[index:index + self.batch_size]
                for index in range(0, len(values), self.batch_size)
            ]
            with ThreadPoolExecutor(max_workers=self.max_concurrency) as executor:
                parts = list(executor.map(self._encode_remote_batch, batches))
            vectors = np.vstack(parts)
        elif self.local_model is not None:
            vectors = self.local_model.encode(
                values,
                batch_size=self.batch_size,
                show_progress_bar=False,
                convert_to_numpy=True,
                normalize_embeddings=False,
            ).astype(np.float32, copy=False)
            self.stats["local_processed_text_count"] += len(values)
        else:
            if self.tokenizer is None or self.transformer is None:
                raise RuntimeError("transformers mean-pooling backend 未初始化")
            parts = []
            with torch.inference_mode():
                for start in range(0, len(values), self.batch_size):
                    tokens = self.tokenizer(
                        values[start:start + self.batch_size],
                        padding=True,
                        truncation=True,
                        max_length=512,
                        return_tensors="pt",
                    ).to(self.device)
                    hidden = self.transformer(**tokens).last_hidden_state
                    mask = tokens["attention_mask"].unsqueeze(-1).to(hidden.dtype)
                    pooled = (hidden * mask).sum(dim=1) / mask.sum(dim=1).clamp_min(1.0)
                    parts.append(pooled.float().cpu().numpy())
            vectors = np.vstack(parts)
            self.stats["local_processed_text_count"] += len(values)
        if vectors.shape != (len(values), self.dimensions):
            raise ValueError(
                f"{self.model} embedding 形状错误: expected={(len(values), self.dimensions)}, actual={vectors.shape}"
            )
        norms = np.linalg.norm(vectors, axis=1, keepdims=True)
        return np.asarray(vectors / np.maximum(norms, 1e-12), dtype=np.float32)


def hpo_surface_rows(surfaces: Sequence[SurfaceForm]) -> list[dict[str, Any]]:
    synonym_ordinals: dict[str, int] = defaultdict(int)
    rows = []
    for index, surface in enumerate(surfaces):
        ordinal = -1
        if surface.synonym_type != "preferred_label":
            ordinal = synonym_ordinals[surface.ontology_id]
            synonym_ordinals[surface.ontology_id] += 1
        rows.append({
            "id": f"surface_{index:07d}",
            "index": index,
            "text": surface.text,
            "hpo_id": surface.ontology_id,
            "preferred_label": surface.preferred_name,
            "surface_type": surface.synonym_type,
            "synonym_ordinal": ordinal,
        })
    return rows


def expected_hpo_collection_metadata(
    spec: HPOEmbeddingSpec,
    terms: dict[str, OntologyTerm],
    surfaces: Sequence[SurfaceForm],
    ontology_hash: str,
) -> dict[str, Any]:
    raw_surface_count = len(terms) + sum(
        1 for term in terms.values() for text, _ in term.synonyms if str(text).strip()
    )
    fingerprint = surface_catalog_fingerprint(surfaces)
    return {
        "ontology": "hpo",
        "ontology_release": "2026-06-23",
        "catalog_version": "hp/releases/2026-06-23",
        "hpo_jsonl_sha256": ontology_hash,
        "surface_manifest_sha256": fingerprint,
        "catalog_fingerprint": fingerprint,
        "concept_count": len(terms),
        "expected_surface_count": len(surfaces),
        "raw_surface_record_count": raw_surface_count,
        "duplicate_surface_record_count": raw_surface_count - len(surfaces),
        "catalog_deduplication": (
            "exact stripped surface text within each HPO concept; preferred label wins "
            "and the first synonym occurrence is retained"
        ),
        "embedding_model": spec.model_name,
        "model_revision": spec.revision,
        "embedding_backend": spec.backend,
        "pooling": spec.pooling,
        "embedding_dimensions": spec.dimension,
        "embedding_batch_size": spec.batch_size,
        "vector_normalization": "L2",
        "distance_metric": "cosine",
        "hnsw_space": "cosine",
        "index_unit": "surface",
        "surface_order": "manifest index",
        "embedding_input": "surface text only",
    }


def hpo_collection_identity_matches(collection: Any, expected: dict[str, Any]) -> bool:
    metadata = collection.metadata or {}
    keys = (
        "hpo_jsonl_sha256", "surface_manifest_sha256", "catalog_fingerprint",
        "expected_surface_count", "embedding_model", "model_revision", "pooling",
        "embedding_dimensions", "vector_normalization", "distance_metric", "hnsw_space",
    )
    if any(str(metadata.get(key)) != str(expected[key]) for key in keys):
        return False
    configuration = collection.configuration_json or {}
    return str((configuration.get("hnsw") or {}).get("space") or "").lower() == "cosine"


def existing_collection_ids(collection: Any, chunk_size: int = 5000) -> set[str]:
    ids: set[str] = set()
    for offset in range(0, collection.count(), chunk_size):
        payload = collection.get(offset=offset, limit=chunk_size, include=[])
        ids.update(str(value) for value in payload["ids"])
    return ids


def verify_hpo_collection_surfaces(
    collection: Any,
    rows: Sequence[dict[str, Any]],
    include_embeddings: bool = False,
) -> str | None:
    digest = hashlib.sha256()
    for start in range(0, len(rows), 512):
        expected_rows = rows[start:start + 512]
        include = ["documents", "metadatas"]
        if include_embeddings:
            include.append("embeddings")
        payload = collection.get(ids=[row["id"] for row in expected_rows], include=include)
        positions = {str(value): index for index, value in enumerate(payload["ids"])}
        for row in expected_rows:
            if row["id"] not in positions:
                raise ValueError(f"{collection.name} 缺少 surface ID: {row['id']}")
            position = positions[row["id"]]
            metadata = payload["metadatas"][position] or {}
            actual = (
                str(payload["documents"][position]),
                str(metadata.get("hpo_id") or metadata.get("ontology_id") or ""),
                int(metadata.get("surface_index", row["index"])),
            )
            expected_value = (row["text"], row["hpo_id"], row["index"])
            if actual != expected_value:
                raise ValueError(
                    f"{collection.name} surface 顺序或文本不一致: expected={expected_value}, actual={actual}"
                )
            if include_embeddings:
                vector = np.asarray(payload["embeddings"][position], dtype=np.float32)
                digest.update(vector.tobytes(order="C"))
    return digest.hexdigest() if include_embeddings else None


def prepare_hpo_model_collection(
    spec: HPOEmbeddingSpec,
    terms: dict[str, OntologyTerm],
    surfaces: Sequence[SurfaceForm],
    ontology_hash: str,
    config: dict[str, Any],
    rebuild: bool = False,
    replace_incompatible: bool = False,
) -> tuple[Any, dict[str, Any]]:
    db_path = (ROOT / "chroma_db_hpo_260623").resolve()
    db_path.mkdir(parents=True, exist_ok=True)
    client = chromadb.PersistentClient(path=str(db_path))
    names = {item if isinstance(item, str) else item.name for item in client.list_collections()}
    expected = expected_hpo_collection_metadata(spec, terms, surfaces, ontology_hash)
    collection = client.get_collection(spec.collection_name) if spec.collection_name in names else None
    if collection is not None and rebuild:
        client.delete_collection(spec.collection_name)
        collection = None
    if collection is not None and not hpo_collection_identity_matches(collection, expected):
        if not replace_incompatible:
            raise ValueError(f"HPO collection 身份不匹配，拒绝覆盖: {spec.collection_name}")
        client.delete_collection(spec.collection_name)
        collection = None
    if collection is None:
        collection = client.create_collection(
            spec.collection_name,
            metadata={**expected, "build_status": "building"},
            configuration={"hnsw": {"space": "cosine"}},
        )

    rows = hpo_surface_rows(surfaces)
    expected_ids = {row["id"] for row in rows}
    existing_ids = existing_collection_ids(collection)
    unexpected_ids = existing_ids - expected_ids
    if unexpected_ids:
        raise ValueError(f"{spec.collection_name} 含有非 catalog ID: {sorted(unexpected_ids)[0]}")
    if existing_ids:
        verify_hpo_collection_surfaces(
            collection, [row for row in rows if row["id"] in existing_ids]
        )
    initial_count = len(existing_ids)
    if initial_count == len(rows) and str((collection.metadata or {}).get("build_status")) == "complete":
        return collection, {
            "model_key": spec.key,
            "model": spec.model_name,
            "revision": spec.revision,
            "pooling": spec.pooling,
            "dimension": spec.dimension,
            "collection_name": collection.name,
            "count": collection.count(),
            "action": "reused",
            "catalog_fingerprint": expected["catalog_fingerprint"],
            "collection_metadata": collection.metadata,
        }

    config_section = spec.config_section or "embedding"
    backend_config = config.get(config_section, {})
    concurrency = int(backend_config.get("max_concurrency", 1))
    embedder = PinnedEmbeddingBackend(spec, backend_config, concurrency)
    started = time.perf_counter()
    for start in range(0, len(rows), spec.batch_size):
        batch_rows = [row for row in rows[start:start + spec.batch_size] if row["id"] not in existing_ids]
        if not batch_rows:
            continue
        vectors = embedder.encode([row["text"] for row in batch_rows])
        collection.add(
            ids=[row["id"] for row in batch_rows],
            documents=[row["text"] for row in batch_rows],
            metadatas=[{
                "ontology_id": row["hpo_id"],
                "hpo_id": row["hpo_id"],
                "preferred_label": row["preferred_label"],
                "surface_type": row["surface_type"],
                "surface_index": row["index"],
                "synonym_ordinal": row["synonym_ordinal"],
                "catalog_fingerprint": expected["catalog_fingerprint"],
            } for row in batch_rows],
            embeddings=vectors.tolist(),
        )
        existing_ids.update(row["id"] for row in batch_rows)
        print(f"{spec.key}: {collection.count()}/{len(rows)} surfaces")
    if collection.count() != len(rows):
        raise RuntimeError(
            f"{spec.collection_name} 构建不完整: expected={len(rows)}, actual={collection.count()}"
        )
    embedding_hash = verify_hpo_collection_surfaces(collection, rows, include_embeddings=True)
    finished_metadata = {
        **expected,
        "build_status": "complete",
        "final_surface_count": len(rows),
        "embedding_matrix_sha256": embedding_hash,
    }
    collection.modify(metadata=finished_metadata)
    result = {
        "model_key": spec.key,
        "model": spec.model_name,
        "revision": spec.revision,
        "pooling": spec.pooling,
        "dimension": spec.dimension,
        "collection_name": collection.name,
        "count": collection.count(),
        "action": "resumed_and_built" if initial_count else "built",
        "initial_count": initial_count,
        "new_vector_count": len(rows) - initial_count,
        "catalog_fingerprint": expected["catalog_fingerprint"],
        "hpo_jsonl_sha256": ontology_hash,
        "embedding_matrix_sha256": embedding_hash,
        "build_seconds": time.perf_counter() - started,
        **embedder.stats,
        "collection_metadata": collection.metadata,
    }
    result_path = PRECOMPUTE_RESULTS_DIR / f"hpo_chroma_build_{spec.key}.json"
    result_path.write_text(
        json.dumps(result, ensure_ascii=False, indent=2) + "\n", encoding="utf-8-sig"
    )
    return collection, result


def build_requested_hpo_collections(config: dict[str, Any]) -> list[dict[str, Any]]:
    requested = [str(value) for value in HPO_CHROMA_BUILD_MODELS]
    if not requested:
        return []
    keys = list(HPO_SPECS_BY_KEY) if "all" in requested else requested
    invalid = sorted(set(keys) - set(HPO_SPECS_BY_KEY))
    if invalid:
        raise ValueError(f"未知 HPO_CHROMA_BUILD_MODELS: {invalid}")
    hpo_terms, hpo_hash = load_ontology("hpo")
    hpo_surfaces = build_surfaces(hpo_terms, "hpo")
    results = []
    for key in keys:
        _, result = prepare_hpo_model_collection(
            HPO_SPECS_BY_KEY[key],
            hpo_terms,
            hpo_surfaces,
            hpo_hash,
            config,
            rebuild=HPO_CHROMA_REBUILD_SELECTED,
            replace_incompatible=HPO_CHROMA_REPLACE_INCOMPATIBLE,
        )
        results.append(result)
        if torch.cuda.is_available():
            torch.cuda.empty_cache()
    if set(keys) == set(HPO_SPECS_BY_KEY):
        fingerprints = {result["catalog_fingerprint"] for result in results}
        counts = {int(result["count"]) for result in results}
        if len(fingerprints) != 1 or counts != {len(hpo_surfaces)}:
            raise RuntimeError("六个 HPO collection 的 surface catalog 不一致")
    summary = {
        "status": "complete",
        "requested_models": keys,
        "surface_count": len(hpo_surfaces),
        "catalog_fingerprint": surface_catalog_fingerprint(hpo_surfaces),
        "results": results,
    }
    (PRECOMPUTE_RESULTS_DIR / "hpo_chroma_build_summary.json").write_text(
        json.dumps(summary, ensure_ascii=False, indent=2) + "\n", encoding="utf-8-sig"
    )
    return results
```

## 5. Dense 检索、检查点与 Excel 输出

```python
# 中文注释：使用单路 OAR cosine 检索生成每个样本的 Top-20 concept。
def retrieve_one(
    sample: dict[str, Any],
    query_vector: np.ndarray,
    dense_collection: Any,
    terms: dict[str, OntologyTerm],
    dataset_spec: dict[str, Any],
) -> dict[str, Any]:
    query = str(sample[dataset_spec["query_col"]])
    candidates = dense_search(dense_collection, query_vector, RETRIEVAL_TOP_K)
    if len(candidates) != RETRIEVAL_TOP_K:
        raise RuntimeError(
            f"sample_id={sample['sample_id']} 只检索到 {len(candidates)} 个唯一 concept，"
            f"要求 {RETRIEVAL_TOP_K}"
        )
    gold_id = str(sample[dataset_spec["gold_col"]])
    accepted_ids = accepted_gold_ids(sample, dataset_spec)
    accepted_set = set(accepted_ids)
    gold_rank = next(
        (rank for rank, item in enumerate(candidates, 1) if item.ontology_id in accepted_set),
        None,
    )
    return {
        "sample_id": int(sample["sample_id"]),
        "source_excel_row": int(sample["source_excel_row"]),
        "query": query,
        "preferred_name": str(sample[dataset_spec["preferred_col"]]),
        "gold_id": gold_id,
        "accepted_gold_ids": list(accepted_ids),
        "metadata": {
            column: str(sample[column]) for column in dataset_spec["metadata_cols"]
        },
        "candidate_ids": [item.ontology_id for item in candidates],
        "candidate_scores": [round(item.dense_score, 10) for item in candidates],
        "candidate_matched_surfaces": [item.matched_surface for item in candidates],
        "gold_candidate_rank": gold_rank,
    }


# 中文注释：读取 JSONL 检查点，只复用查询、Gold 和 accepted IDs 都一致的 Dense 结果。
def load_checkpoint(
    path: Path,
    samples_by_id: dict[int, dict[str, Any]],
    dataset_spec: dict[str, Any],
) -> dict[int, dict[str, Any]]:
    completed: dict[int, dict[str, Any]] = {}
    if not RESUME or not path.exists():
        return completed
    with path.open("r", encoding="utf-8-sig") as handle:
        for line in handle:
            if not line.strip():
                continue
            item = json.loads(line)
            sample = samples_by_id.get(int(item.get("sample_id", -1)))
            if sample is None:
                continue
            if item.get("query") != str(sample[dataset_spec["query_col"]]):
                continue
            if item.get("gold_id") != str(sample[dataset_spec["gold_col"]]):
                continue
            if item.get("accepted_gold_ids") != list(accepted_gold_ids(sample, dataset_spec)):
                continue
            if (
                len(item.get("candidate_ids", [])) == RETRIEVAL_TOP_K
                and len(item.get("candidate_scores", [])) == RETRIEVAL_TOP_K
                and len(item.get("candidate_matched_surfaces", [])) == RETRIEVAL_TOP_K
            ):
                completed[int(item["sample_id"])] = item
    return completed


# 中文注释：追加写入一批检查点记录，保证中断后不重复计算 query embedding 和 Dense 检索。
def append_checkpoint(path: Path, records: Sequence[dict[str, Any]]) -> None:
    path.parent.mkdir(parents=True, exist_ok=True)
    with path.open("a", encoding="utf-8") as handle:
        for record in records:
            handle.write(json.dumps(record, ensure_ascii=False, separators=(",", ":")) + "\n")


# 中文注释：为大表添加筛选、冻结窗格、列宽和 Excel 表格样式。
def style_sheet(workbook_path: Path, sheet_name: str, widths: dict[str, float]) -> None:
    workbook = load_workbook(workbook_path)
    sheet = workbook[sheet_name]
    sheet.freeze_panes = "A2"
    sheet.sheet_view.showGridLines = False
    for column, width in widths.items():
        sheet.column_dimensions[column].width = width
    for cell in sheet[1]:
        cell.fill = PatternFill("solid", fgColor="176567")
        cell.font = Font(color="FFFFFF", bold=True)
        cell.alignment = Alignment(horizontal="center", vertical="center", wrap_text=True)
    if sheet.max_row >= 2 and sheet.max_column >= 1:
        table = Table(displayName=f"{sheet_name.replace('_', '')}Table", ref=sheet.dimensions)
        table.tableStyleInfo = TableStyleInfo(
            name="TableStyleMedium2",
            showFirstColumn=False,
            showLastColumn=False,
            showRowStripes=True,
            showColumnStripes=False,
        )
        sheet.add_table(table)
    workbook.save(workbook_path)


# 中文注释：把 Dense Top-20 压缩成单样本一行，并仅汇报 Recall@1。
def build_output_frame(
    records: Sequence[dict[str, Any]],
    terms: dict[str, OntologyTerm],
    dataset_spec: dict[str, Any],
) -> pd.DataFrame:
    rows: list[dict[str, Any]] = []
    for record in sorted(records, key=lambda item: int(item["sample_id"])):
        row: dict[str, Any] = {
            "sample_id": record["sample_id"],
            "source_excel_row": record["source_excel_row"],
            "query": record["query"],
            "preferred_name": record["preferred_name"],
            "gold_id": record["gold_id"],
            "accepted_gold_ids_json": json.dumps(
                record["accepted_gold_ids"], ensure_ascii=False, separators=(",", ":")
            ),
            "gold_candidate_rank": record["gold_candidate_rank"],
            "hit_at_1": bool(
                record["gold_candidate_rank"] and record["gold_candidate_rank"] <= 1
            ),
            "candidate_ids_json": json.dumps(
                record["candidate_ids"], ensure_ascii=False, separators=(",", ":")
            ),
            "candidate_scores_json": json.dumps(
                record["candidate_scores"], separators=(",", ":")
            ),
            "candidate_matched_surfaces_json": json.dumps(
                record["candidate_matched_surfaces"],
                ensure_ascii=False,
                separators=(",", ":"),
            ),
        }
        row.update(record["metadata"])
        for rank, ontology_id in enumerate(record["candidate_ids"][:20], 1):
            row[f"top_{rank:02d}_id"] = ontology_id
            row[f"top_{rank:02d}_name"] = terms[ontology_id].name
        rows.append(row)
    return pd.DataFrame(rows)


# 中文注释：写出不含 Summary 的 Dense 预计算工作簿，并记录可复现参数。
def export_precompute(
    output_path: Path,
    output_frame: pd.DataFrame,
    run_config: dict[str, Any],
) -> None:
    config_frame = pd.DataFrame([
        {
            "parameter": key,
            "value": json.dumps(value, ensure_ascii=False)
            if isinstance(value, (dict, list))
            else value,
        }
        for key, value in run_config.items()
    ])
    with pd.ExcelWriter(output_path, engine="openpyxl") as writer:
        output_frame.to_excel(writer, sheet_name="Samples", index=False)
        config_frame.to_excel(writer, sheet_name="Run_Config", index=False)
    style_sheet(
        output_path,
        "Samples",
        {"A": 12, "B": 16, "C": 42, "D": 32, "E": 18, "F": 60, "G": 18},
    )
    style_sheet(output_path, "Run_Config", {"A": 36, "B": 100})
```

## 6. 执行 OAR 全流程

运行本单元会产生 query embedding/API 费用；已有 OAR 索引和与当前配置匹配的检查点会自动复用。

```python
# 中文注释：解析参数、加载术语与测试集，并确保 Top-20 设置有效。
if DATASET_KEY not in DATASETS:
    raise KeyError(f"未知 DATASET_KEY: {DATASET_KEY}")
if RETRIEVAL_TOP_K != 20:
    raise ValueError("本项目预计算固定保存 Top-20，请保持 RETRIEVAL_TOP_K=20")

config = load_config(CONFIG_PATH)
dataset_frame, dataset_spec = load_samples(DATASET_KEY, MAX_SAMPLES)
ontology_key = dataset_spec["ontology"]
terms, ontology_hash = load_ontology(ontology_key)
surfaces = build_surfaces(terms, ontology_key)

# 中文注释：可选地在主 notebook 内构建/校验六个固定 HPO Chroma collection。
hpo_chroma_build_results = build_requested_hpo_collections(config)

# 中文注释：六个固定模型严格复用构建时的 revision 和 pooling；其他自定义模型沿用通用后端。
resolved_embedding_model = str(EMBEDDING_MODEL)
pinned_spec = HPO_SPECS_BY_MODEL.get(resolved_embedding_model)
if pinned_spec is not None:
    if EMBEDDING_DIMENSIONS not in (None, pinned_spec.dimension):
        raise ValueError(
            f"固定模型 {resolved_embedding_model} 的维度必须为 {pinned_spec.dimension}"
        )
    embedding_config = config.get(pinned_spec.config_section or "embedding", {})
    resolved_embedding_backend = pinned_spec.backend
    resolved_dimensions = pinned_spec.dimension
    resolved_concurrency = int(
        EMBEDDING_MAX_CONCURRENCY or embedding_config.get("max_concurrency", 1)
    )
    embedder = PinnedEmbeddingBackend(pinned_spec, embedding_config, resolved_concurrency)
else:
    embedding_config = config["embedding"]
    if EMBEDDING_IS_LOCAL is None:
        resolved_embedding_backend = (
            "openai" if resolved_embedding_model.startswith("text-embedding-")
            else "sentence_transformers"
        )
    else:
        resolved_embedding_backend = "sentence_transformers" if EMBEDDING_IS_LOCAL else "openai"
    resolved_dimensions = EMBEDDING_DIMENSIONS
    resolved_concurrency = int(
        EMBEDDING_MAX_CONCURRENCY or embedding_config.get("max_concurrency", 1)
    )
    embedder = EmbeddingBackend(
        resolved_embedding_backend,
        resolved_embedding_model,
        embedding_config,
        resolved_dimensions,
        resolved_concurrency,
        EMBEDDING_BATCH_SIZE,
    )

# 中文注释：初始化或复用唯一的 Dense 索引；主流程不建立词法通道。
print(f"dataset={DATASET_KEY}; samples={len(dataset_frame)}; ontology={ontology_key}; terms={len(terms)}; surfaces={len(surfaces)}")
use_oar = bool(USE_OAR_HPO_RETRIEVAL and ontology_key == "hpo" and resolved_embedding_model == "text-embedding-3-large")
oar_runtime = load_default_oar_runtime(ROOT) if use_oar else None
if oar_runtime is not None:
    dense_collection, dense_collection_info = prepare_oar_collection(surfaces, oar_runtime)
elif ontology_key == "hpo" and pinned_spec is not None:
    dense_collection, dense_collection_info = prepare_hpo_model_collection(
        pinned_spec,
        terms,
        surfaces,
        ontology_hash,
        config,
        rebuild=REBUILD_DENSE_INDEX,
        replace_incompatible=HPO_CHROMA_REPLACE_INCOMPATIBLE,
    )
    dense_collection_info = {
        **dense_collection_info,
        "chroma_db_dir": str((ROOT / "chroma_db_hpo_260623").resolve()),
        "dense_collection_name": dense_collection.name,
        "dense_collection_count": dense_collection.count(),
        "dense_index_unit": "surface",
        "dense_collection_action": dense_collection_info["action"],
        "dense_catalog_version": "hp/releases/2026-06-23",
    }
else:
    dense_collection, dense_collection_info = prepare_dense_collection(
        ontology_key, ontology_hash, surfaces, embedder
    )

# 中文注释：用完整检索配置生成短哈希，既隔离不同运行又避免 Windows 路径过长。
model_tag = safe_tag(f"{resolved_embedding_model}_oar") if use_oar else safe_tag(resolved_embedding_model)
checkpoint_identity = {
    "pipeline_version": 3,
    "retrieval_method": "oar_dense_cosine" if use_oar else "dense_cosine",
    "dataset_key": DATASET_KEY,
    "ontology": ontology_key,
    "ontology_sha256": ontology_hash,
    "embedding_backend": resolved_embedding_backend,
    "embedding_model": resolved_embedding_model,
    "embedding_dimensions": resolved_dimensions,
    "dense_collection_name": dense_collection_info["dense_collection_name"],
    "dense_collection_count": dense_collection_info["dense_collection_count"],
    "dense_catalog_version": dense_collection_info["dense_catalog_version"],
    "dense_index_unit": dense_collection_info["dense_index_unit"],
    "oar_enabled": use_oar,
    "oar_run_id": oar_runtime.run_id if oar_runtime is not None else None,
    "oar_model_sha256": oar_runtime.model_sha256 if oar_runtime is not None else None,
    "retrieval_top_k": RETRIEVAL_TOP_K,
}
checkpoint_payload = json.dumps(checkpoint_identity, ensure_ascii=True, sort_keys=True, separators=(",", ":"))
checkpoint_config_hash = hashlib.sha256(checkpoint_payload.encode("utf-8")).hexdigest()[:16]
checkpoint_path = (
    CACHE_DIR / "retrieval_checkpoints"
    / f"{DATASET_KEY}_{model_tag}_{checkpoint_config_hash}.jsonl"
)
samples = dataset_frame.to_dict(orient="records")
samples_by_id = {int(sample["sample_id"]): sample for sample in samples}
completed = load_checkpoint(checkpoint_path, samples_by_id, dataset_spec)
pending = [sample for sample in samples if int(sample["sample_id"]) not in completed]
print(f"completed={len(completed)}; pending={len(pending)}; embedding_concurrency={resolved_concurrency}")

# 中文注释：成批计算 query embedding，执行单路 Dense 检索并立即追加检查点。
started_at = time.perf_counter()
for start in range(0, len(pending), EMBEDDING_BATCH_SIZE):
    batch = pending[start:start + EMBEDDING_BATCH_SIZE]
    queries = [str(sample[dataset_spec["query_col"]]) for sample in batch]
    vectors = embedder.encode(queries)
    if oar_runtime is not None:
        vectors = oar_runtime.project(vectors)
    batch_records = [
        retrieve_one(sample, vector, dense_collection, terms, dataset_spec)
        for sample, vector in zip(batch, vectors)
    ]
    append_checkpoint(checkpoint_path, batch_records)
    completed.update({int(record["sample_id"]): record for record in batch_records})
    print(f"retrieval: {len(completed)}/{len(samples)}")

# 中文注释：构造单样本一行的 Dense Top-20 输出，但方法指标只汇报 Recall@1。
final_records = [completed[int(sample["sample_id"])] for sample in samples]
output_frame = build_output_frame(final_records, terms, dataset_spec)
recall = {"recall_at_1": float(output_frame["hit_at_1"].mean())}
output_path = PRECOMPUTE_RESULTS_DIR / f"{DATASET_KEY}_retrieval_top{RETRIEVAL_TOP_K}_{model_tag}_dense-only.xlsx"
run_config = {
    "dataset_key": DATASET_KEY,
    "dataset_path": str(dataset_spec["path"]),
    "dataset_sheet": dataset_spec["sheet"],
    "accepted_gold_policy": "current ontology ID plus official accepted historical IDs",
    "sample_count": len(output_frame),
    "ontology": ontology_key,
    "ontology_release": ONTOLOGIES[ontology_key]["release"],
    "ontology_sha256": ontology_hash,
    "embedding_backend": resolved_embedding_backend,
    "embedding_model": resolved_embedding_model,
    "embedding_dimensions": resolved_dimensions,
    "embedding_max_concurrency": resolved_concurrency,
    "retrieval_top_k": RETRIEVAL_TOP_K,
    "retrieval_method": "oar_dense_cosine" if use_oar else "dense_cosine",
    "checkpoint_path": str(checkpoint_path),
    "checkpoint_config_hash": checkpoint_config_hash,
    "checkpoint_identity": checkpoint_identity,
    "elapsed_seconds_this_run": round(time.perf_counter() - started_at, 3),
    **dense_collection_info,
    **recall,
}
export_precompute(output_path, output_frame, run_config)
print(f"saved={output_path}")
print(recall)
```
